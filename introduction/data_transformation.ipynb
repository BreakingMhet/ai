```python
# Data Transformation
## Encoding: converts categorical values into numeric form so machine learning algorithms can utilize these features.
```

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("../datasets/demo_dataset.csv")
print(df['protocol'].unique())

# One-hot encoding: Takes a categorical feature and converts it into a set of new binary features, where each binary feature corresponds to one possible category value.
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
encoded = encoder.fit_transform(df[['protocol']])

encoded_df = pd.DataFrame(encoded, columns=encoder.get_feature_names_out(['protocol']))
df = pd.concat([df.drop('protocol', axis=1), encoded_df], axis=1)

print("\n**************AFTER Encoding*************")
print(df.columns.tolist())
```
